```python
from pyspark import pipelines as dp

@dp.materialized_view(name="fct_results")
def fct_results():
    sample = [
        (1, 101, 1001, 12000, 42.2, 35),
        (2, 102, 1002, 11000, 42.2, 29),
    ]
    columns = [
        "result_id",        # integer
        "event_id",         # integer
        "athlete_id",       # integer
        "performance_seconds", # integer
        "performance_km",      # float
        "athlete_age"           # integer
    ]
    return spark.createDataFrame(sample, columns)
```
